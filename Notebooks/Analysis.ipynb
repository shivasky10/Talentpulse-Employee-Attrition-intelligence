#  SQL Workforce Analytics

## Objective

SQL was used to perform workforce analytics and extract meaningful business insights from employee data.

The cleaned employee dataset was exported to CSV and imported into a MySQL database named `TalentPulse`, with a table named `employee_attrition`.

The analysis focused on understanding employee attrition patterns across departments, job roles, overtime status, salary levels and employee experience.

## Business Questions

The following business questions were investigated using SQL:

1. What is the overall employee attrition rate?
2. Which departments have the highest attrition rates?
3. Which job roles experience the highest employee turnover?
4. How is overtime associated with employee attrition?
5. How do salary, total working experience and company tenure differ between employees who leave and those who stay?

## SQL Techniques Used

- SELECT and WHERE
- GROUP BY and ORDER BY
- Aggregate functions: COUNT(), SUM(), AVG()
- CASE WHEN conditional expressions
- ROUND() for numerical formatting
- Conditional aggregation

## Key Findings

The SQL analysis was used to identify departments and job roles with higher attrition rates, examine the relationship between overtime and attrition, and compare employee characteristics across retention groups.

These findings complement the machine learning analysis and help translate data into actionable workforce insights.

# Business Insights and Recommendations

## 1 Key Business Insights

Based on the exploratory data analysis and SQL workforce analytics, the following patterns were identified:

**1. Overall Employee Attrition**

The dataset contains 1,470 employees, of which 237 left the organization. The overall attrition rate is approximately 16.12%.

**2. Job Role-wise Attrition**

Sales Representatives have the highest observed attrition rate at approximately 39.76%, followed by Laboratory Technicians at approximately 23.94%.

This suggests that certain job roles may require closer retention monitoring.

**3. Overtime and Attrition**

Employees working overtime have an observed attrition rate of approximately 30.53%, compared with 10.44% among employees not working overtime.

This indicates a strong association between overtime and employee attrition.

**4. Employee Experience and Tenure**

Employees who left had lower average total working experience and shorter company tenure than employees who stayed.

- Average total working years: 8.24 years for employees who left versus 11.86 years for employees who stayed.
- Average company tenure: 5.13 years for employees who left versus 7.37 years for employees who stayed.

**5. Salary Differences**

Employees who left had a lower average monthly income (approximately 4,787) than employees who stayed (approximately 6,833).

These differences may indicate potential relationships between compensation, experience and employee retention.

## 2 Business Recommendations

Based on the observed patterns, the following actions could be considered:

**1. Targeted Retention Strategies**

HR teams could investigate retention challenges in roles with higher attrition, particularly Sales Representatives and Laboratory Technicians.

**2. Overtime Management**

Organizations could review overtime distribution, workload allocation and employee well-being to identify potential retention risks.

**3. Early-Career Employee Support**

Employees with shorter tenure could benefit from structured onboarding, mentoring and career development opportunities.

**4. Compensation Review**

HR teams could examine compensation differences across job roles, experience levels and performance groups to identify potential retention concerns.

**5. Predictive Attrition Monitoring**

The XGBoost model could be used as a decision-support tool to identify employees with elevated predicted attrition risk. HR professionals should review predictions alongside employee feedback and other relevant context.
